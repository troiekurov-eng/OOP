```python
class LinearRegressionModel():

    def __init__(self, data, x_units: str, y_units: str): # инициализация объекта (экземпляра)
        self.data = data # загрузка данных
        self.trained = False # признак тренировки и готовности модели
        self.slope = None # наклон модели a 
        self.intersection = None # пересечение модели b
        # единицы измерения осей
        self.x_units = x_units 
        self.y_units = y_units

    def train(self):
        x = [point[0] for point in self.data]  # point = (#,#) точнка (пара)
        y = [point[1] for point in self.data]
        n = len(self.data)
        #формула
        x_mean = sum(x) / n  # <x>
        y_mean = sum(y) / n  # <y>
        xy = [x[i] * y[i] for i in range(n)]
        xy_mean = sum(xy) / n  # <xy>
        x_mean_squared = pow(x_mean, 2)  # <x>^2
        x2 = [value**2 for value in x]  # x^2
        x_squared_mean = sum(x2) / n  # <x^2>
        b = (xy_mean - x_mean * y_mean) / (x_squared_mean - x_mean_squared) #наклон
        a = y_mean - b*x_mean #пересечение
        #запоминание результата обучения
        self.slope = b
        self.intersection = a
        self.trained = True
        return a, b       

    def plot(self, title):
        fig, ax = plt.subplots()
        x = [point[0] for point in self.data]
        y = [point[1] for point in self.data]
        ax.scatter(x, y, marker = 'x')
        ax.set_title(title)
        ax.set_xlabel(self.x_units)
        ax.set_ylabel(self.y_units)
        return fig, ax
```
