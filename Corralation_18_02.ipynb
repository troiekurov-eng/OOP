```python
n = 100
```

```python
ages = np.random.normal(40, 15, size=n).round(0)
```

```python
ages
```

```python
mass = np.random.normal(70, 20, size=n).round(0)
mass
```

```python
import matplotlib.pyplot as plt
```

Интерес к данным в "паре".

Co-relation  Со-отношение "параллельное" отношение

- Корреляция — это статистическая мера, показывающая, насколько две переменные связаны между собой. Она помогает выявлять зависимости, но не причинно-следственные связи (корреляция ≠ causation).

Базовая визуализация для пары это scatter plot 

```python
am = pd.DataFrame({
    'age': ages,
    'mass': mass,
})
am
```

```python
sns.scatterplot(data=am, x='age', y='mass')
```

```python
tall = np.random.normal(170, 50, size=n).round(0)
tall
```

```python
mass = (30*20*tall)/1000
mass
```

```python
df = pd.DataFrame({
    'tall': tall,
    'mass': mass,
})
df
```

```python
sns.scatterplot(data=df, x='tall', y='mass')
```

- **Корреляция Пирсона (Pearson correlation):** Измеряет линейную связь между непрерывными переменными. Коэффициент r от -1 до 1. Требует нормального распределения данных.

- 0 - нет корреляции,
- 1 - положительная корреляция (корреляция)
- -1 - отрицательная корреляция (антикорреляция)

```python
am.corr(method = 'pearson')
```

```python
df.corr(method = 'pearson')
```

```python
years_left = (100 * tall.max() - tall)/((tall.max() - tall.min())).round(0)
years_left
```

```python
yf = pd.DataFrame({
    'tall': tall,
    'years_left': years_left,
})
yf
```

```python
sns.scatterplot(data=yf, x='tall', y='years_left')
```

```python
yf.corr(method = 'pearson')
```

```python
wide = np.random.normal(20, 5, size=n).round(0)
wide
```

```python
mass2 = (wide*wide*tall)/1000
mass2
```

```python
df2 = pd.DataFrame({
    'tall': tall,
    'mass2': mass2,
})
df2
```

```python
#sns.scatterplot(data=yf, x='tall', y='mass2')
```

```python
for i in range(0, 10+1):
    s = 2*(i-10)**3
   # print(s/10)
    wide = np.random.normal(20, i, size=n).round(0)
    #print('std', wide.std())
    mass_w = (wide*wide*tall)/1000
    tf = pd.DataFrame({
    'tall': tall,
    'mass': mass_w,
    })
    cm = tf.corr(method = 'pearson')
    pearson_coef = cm.loc['tall', 'mass'].round(0)
    #print('corr', pearson_coef)
    sns.scatterplot(data=tf, x='tall', y='mass')
    plt.title(f"person {pearson_coef}")
    plt.gcf().set_size_inches(3, 2)
    plt.show()
```
