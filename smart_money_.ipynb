```python
class Money(float):

    valid_currency = {'BYN', 'USD', 'EUR'}
    currency = None
    expired = False

    @classmethod
    def is_expired(cls, obj):
        if not isinstance(obj, cls):
            print('This is not Money')
        else:
            print(f"expired={obj.expired}")
    
    @classmethod
    def is_valid(cls, currency):
        if currency in cls.valid_currency:
            return True
        else:
            return False

    def check_currency(self):
        if self.currency is None:
            print('No currecy is set')
            return False
        elif self.currency in self.valid_currency:
            return True
        else:
            print('Not a valid currency')
            return False

    # method 
    def set_currency(self, currecy: str = 'BYN'):
        if not Money.is_valid(currecy):
            self.check_currency()
            return self 
        self.currency = currecy
        return self

    def print(self):
        print(f"{self} {self.currency}")


    def expire(self):
        self.expired = True
```
